```python
import pandas as pd

products = pd.read_csv("../data/raw/olist_products_dataset.csv")
products.head()
```

```python
products = products.drop(
    columns=[
        "product_name_lenght",
        "product_description_lenght",
        "product_photos_qty",
    ]
)
```

```python
products.info()
```

```python
products.duplicated().sum()
```

```python
measurement_columns = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm",
]
```

```python
(products[measurement_columns] <= 0).sum()

products.loc[
    products["product_weight_g"] <= 0,
    ["product_id", "product_category_name"] + measurement_columns
]
```

```python
products["flag_invalid_weight"] = products["product_weight_g"] <= 0

products.loc[
    products["flag_invalid_weight"],
    "product_weight_g"
] = float("nan")
```

```python
products_with_weight = products.loc[
    products["product_weight_g"].notna()
].copy()
products["product_weight_g"].isna().sum()
```

```python
products.to_csv("../data/processed/products_clean.csv", index=False)
```
